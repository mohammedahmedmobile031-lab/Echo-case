Bronze: The Logic Behind Recovering the Message
How Majority Voting Works: The 96-bit stream is divided into 32 triplets (groups of 3 bits). Because each original bit was sent three times, the transmission process could introduce at most one error per triplet.

The Decision Rule: You look at each triplet and apply a majority rule. If a triplet contains two or more 1s (e.g., 110, 101, 111), you trust that the original bit was a 1. If it contains two or more 0s (e.g., 000, 001, 010), you trust that the original bit was a 0.

The Result: Applying this across all 32 triplets strips out the single-bit corruption errors, revealing the decoded binary message that spells out the word and note.


Silver:
The Breaking Point: The repetition code relies entirely on the assumption that at most one bit in a triplet gets corrupted. If two or more bits in a single triplet flip during transmission, the majority vote is actively misled by the errors and outputs the wrong value.
